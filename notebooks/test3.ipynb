```python
import geopandas as gpd
from shapely.geometry import box

from resplotlib import rpc
```

```python
# Define bounds for the Netherlands
bounds = [3.31497114423, 50.80372101501, 7.09205325687, 53.5104033476]
bounds_crs = "EPSG:4326"
gdf_bounds = gpd.GeoDataFrame(geometry=[box(*bounds)], crs=bounds_crs).to_crs("EPSG:32631")
bounds_utm = gdf_bounds.geometry.total_bounds
gdf_poly = gdf_bounds.copy()
gdf_poly["geometry"] = gdf_poly.geometry.buffer(-100000).buffer(100000)

m = rpc.Map(bounds=bounds_utm, draw_bounds=True, bounds_crs="EPSG:32631", clear_on_draw=True)
m = rpc.explore_basemap(m=m, style="satellite")
m
```
